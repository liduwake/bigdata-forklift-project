# Exercise 3
Three persons work independently to decipher a message. The probabilities that they will deciphe the message are 1/5, 1/4 and 1/3. Calculate the probability that the message will be deciphered? Verify you calculation by simulation.
(Ans. 0.60)

#Solution here
**Solution:**

Since the three persons work independently, So they are indepence.

1 minus the situation of all fails


$
P = 1 - P(\text{all fail})
$

$
= 1 - (1 - \tfrac{1}{5})(1 - \tfrac{1}{4})(1 - \tfrac{1}{3})
$

$
= 1 - \left(\tfrac{4}{5} \cdot \tfrac{3}{4} \cdot \tfrac{2}{3}\right)
= 1 - \tfrac{2}{5}
= \tfrac{3}{5} = 0.6
$

```python
from numpy.random import default_rng

rng = default_rng()

n_trials = 100000
count = 0

for i in range(n_trials):
    p1 = rng.choice(['success', 'fail'], p=[1/5, 4/5])
    p2 = rng.choice(['success', 'fail'], p=[1/4, 3/4])
    p3 = rng.choice(['success', 'fail'], p=[1/3, 2/3])

    if p1 == 'success' or p2 == 'success' or p3 == 'success':
        count += 1

print("Simulated probability =", count / n_trials)
print("Exact probability =", 0.6)
```
