```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, explode, count, avg, row_number, to_date, regexp_replace, lower, length, split, trim
from pyspark.sql.window import Window

spark = SparkSession.builder.appName("YelpDatasetProcessing").getOrCreate()
```

```python
business_df = spark.read.json("yelp-data/yelp_academic_dataset_business.json")
```

```python
flattened_df = business_df.select(
    "business_id",
    "name",
    "stars",
    "review_count",
    "city",
    "state",
    "attributes.WiFi",
    "categories"
)

exploded_df = flattened_df.withColumn(
    "category",
    explode(split("categories", ",\\s*"))
).withColumn(
    "category",
    trim(col("category"))
)
```
