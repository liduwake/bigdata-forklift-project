# Exercise 1

A shooter hits the target in a single shot with the probability of 0.95. Find the probability that in the next two shots the shooter succeeds to hit the target

a. twice, 

b. at least once,

c. exactly once.

(Ans. a. 0.9025, b. 0.9975, 0.095)

**a. twice (independent events)**
$
P = 0.95 \times 0.95 = 0.9025
$


**b. at least once (complement rule)**

**1 minus the probability of missing both**

$
P = 1 - (0.05)^2 = 0.9975
$

**c. exactly once**

**Case 1: first shot succeeds, second fails, $0.95 \times 0.05$**

**Case 2: first shot fails, second succeeds, $0.05 \times 0.95$**

**Add them**

$
P = (0.95 \times 0.05) + (0.05 \times 0.95) = 0.095
$
