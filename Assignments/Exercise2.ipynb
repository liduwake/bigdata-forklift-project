# Exercise 2
10 marbles are in a bag. 3 are red, 4 blue and 3 yellow.

Two marbles are drawn without replacement. Calculate the probability that colors are red and blue. (Order does not matter)


(Ans. 0.27 )

#Solution here


Case 1: red then blue  
$
P_1 = \frac{3}{10} \times \frac{4}{9}
$

Case 2: blue then red  
$
P_2 = \frac{4}{10} \times \frac{3}{9}
$

$
P = P_1 + P_2 
= \frac{3}{10}\cdot\frac{4}{9} + \frac{4}{10}\cdot\frac{3}{9}
= \frac{24}{90} = \frac{4}{15} \approx 0.27
$
