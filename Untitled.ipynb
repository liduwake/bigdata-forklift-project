```python
import pandas as pd
import numpy as np

np.random.seed(42)

data = {
    "timestamp": pd.date_range(start="2024-01-01", periods=1000, freq="min"),
    "device_id": np.random.choice(["forklift_1", "forklift_2", "forklift_3"], 1000),
    "temperature": np.random.normal(25, 5, 1000),
    "vibration": np.random.normal(0.5, 0.2, 1000),
    "status": np.random.choice(["idle", "working", "error"], 1000, p=[0.6, 0.35, 0.05])
}

df = pd.DataFrame(data)
df.head()
```

```python
# 添加异常数据
df.loc[10:20, "temperature"] = 100  # 异常温度

# 清洗
df = df[df["temperature"] < 80]

# 转时间格式
df["timestamp"] = pd.to_datetime(df["timestamp"])

df.head()
```

```python
# 每个设备平均温度
print(df.groupby("device_id")["temperature"].mean())

# 状态分布
print(df["status"].value_counts())
```

```python
spark_df = spark.createDataFrame(df)
spark_df.show(5)
```

```python
spark_df.groupBy("device_id").avg("vibration").show()
```

```python
spark_df.filter(spark_df.status == "error").count()
```

```python
from pyspark.sql.functions import hour

spark_df = spark_df.withColumn("hour", hour("timestamp"))

spark_df.groupBy("hour").count().orderBy("hour").show()
```

```python
pdf = spark_df.toPandas()

import matplotlib.pyplot as plt

pdf.groupby("hour").size().plot(kind="line")
plt.title("Events per Hour")
plt.show()
```
