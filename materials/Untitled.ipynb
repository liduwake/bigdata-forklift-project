```python
import numdifftools as nd
# Define the function
def f(x):
    return x[0]**2 + x[1]**2 # f(x, y) = x^2 + y^2
# Create a gradient object
gradient = nd.Gradient(f)
# Evaluate the gradient at the point (x, y) = (1, 2)
point = [1, 2]
gradient_at_point = gradient(point)
print("Gradient at (1, 2):", gradient_at_point)
```
