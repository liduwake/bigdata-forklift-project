### Calculated values


Sometimes array element value depends on its indices `i` and `j`. The the array can be built using two for loops over columns and rows.

So called [Lehmer matrix](https://en.wikipedia.org/wiki/Lehmer_matrix) is defined by the equation $ A_{ij} = \frac{\min(i,j)}{\max(i,j)}, i,j = 1,2,...,n $
In the formula indices start from 1, but in python they start from 0, so a shift by 1 is needed.



````python
from numpy import zeros
def lehmer(n):
    A = zeros((n,n))
    for i in range(n):
        for j in range(n):
            A[i,j] = min(i + 1, j + 1)/max(i + 1, j + 1)
    return A

print(lehmer(4))            
````

This can be done using the function `numpy.fromfunction`. The a `lambda` function is used to define the formula.

````python
np.fromfunction(lambda i, j: i + j, (2, 3), dtype=int)
````

````python
A = np.fromfunction(lambda i, j: np.minimum(i + 1, j + 1)/np.maximum(i + 1, j + 1), (4, 4), dtype=int)
print(A)
````

Notice that in the `lamba`above `i` and `j` are arrays and python built-in functions `min` and `max` don't return minumum or maximum of an array, but a `ValueError`. Therefore `np.minimum` and `np.maximum` were used.

````python
i = np.random.randint(low = 1, high = 12, size = (4,1))
j = np.random.randint(low = 1, high = 12, size = (4,1))
print("i = ", i)
print("j = ", j)
print("min(i,j) =", np.minimum(i,j))
print(min(i,j)) #Gives an error

````

### Building larger arrays from smaller ones

Sometimes sevaral arrays are joined together to form a larger array. This operation is called **concatenation** or **stacking** of arrays.

The axis keyword is often used in connection with arrays. 1 dimensional arrays only have axis 0. Two dimensional arrays have axes 0 and 1 and so on.

![img](https://i.imgur.com/Rzz9X8B.png)

````python
a = np.array([1,2,3,4])
b = np.array([8,9,10,10])

c = np.concatenate((a,b),axis=0) #in this case same as np.hstack((a,b))
print("c:")
print(c)
d = np.vstack((a,b)) #stack vertically, must have the same number of elements
print("d:")
print(d)
````

This also works for 2 dimensional arrays:

````python
a = np.array([[2,3],[5,6]])
b = np.array([[9,10],[11,12]])

c = np.hstack((a,b))
print(f"Vertical stack of a and b:\n {c}")
d = np.vstack((a,b))
print(f"Horizontal stack of a and b:\n {d}")
print(f"Concatenation with respect to axis 0:\n {np.concatenate((a,b),axis = 0)}")
print(f"Concatenation with respect to axis 1:\n {np.concatenate((a,b),axis = 1)}")
print(f"Transpose of c is \n {np.transpose(c)}")
````

The last operation is the transpose of a matrix. Then its rows are written as columns. The transpose of a $m \times n$ matrix is a $n \times m$. Mathematical notation for the transpose operation of a matrix $A$ is $$A^{T}$$.

### Solving a linear system of equations using matrices

Linear system of equations has $n$ equations in $n$ variables and the solution must satisfy all equations simultaneously.

Three possible cases exist:
* a unique solution can be found, this is the "ordinary case"
* no solution exists, this means that equations are in conflict
* infinite number of solutions , some equation is a combination of one ore more of the other equations

An example of three simultaneous equations in three variables is the following: $$\begin{cases} x_1 + 5x_2 + x_3 = 24 \\ 7x_1 + 1/5x_2 + 4x_3 = 0 \\ x_1 + 3x_2 + 2x_3 = 12 \end{cases} $$

The solution of the system of equations is $$\begin{cases} x_1 = 1 \\ x_2 = 5 \\ x_3 = -2 \end{cases}$$

The solution can be verified by substitution. 

````python
x1 = 1
x2 = 5
x3 = -2

print(x1 + 5*x2 + x3)
print(7*x1 + 1/5*x2 + 4*x3)
print(x1 + 3*x2 + 2*x3)
````

To solve the system using matrices in python the linear system of equations is written in the form $$ \bold A \bold x = \bold b, $$ where $\bold A$ is a square matrix of coefficients and $\bold b$ is the vector of constants in the right hand side of the equations. The vector $\bold x$ is the solution vector. 

The variables $x_1$, $x_2$ and $x_3$ must be in the same order in all three equations. The order of the three equations is irrelevant. 

The first step is to collect a matrix from the coefficients  in the system:
$$\begin{cases} x_1 + 5x_2 + x_3 = 24 \\ 7x_1 + \frac{1}{5}x_2 + 4x_3 = 0 \\ x_1 + 3x_2 + 2x_3 = 12 \end{cases} $$
This gives $$A = \begin{bmatrix}1&5&1\\7&1/5&4\\1&3&2 \end{bmatrix}, b = \begin{bmatrix}24\\0\\12 \end{bmatrix}$$


````python
A = np.array([[1,5,1],[7,1/5,4],[1,3,2]])
b = np.array([24,0,12])
````

````python
xyz = np.linalg.solve(A,b)
print(xyz)
````

### Matrix algebra

Another name for a two dimensional array is **matrix**.
Matrix sum and difference are elementwise operations. This means that elements in same position are either added or subtracted. 

In ```numpy``` there is no need to write loops over all elements of an array to calculate sums of differences.

````python
a = np.array([[2,43],[4,-3]])
b = np.array([[-2,32],[22,33]])

print("a + b = \n", a+b)
print("\na - b = \n", a-b)
````

**Matrix product** is calculated in the following way:

$$\begin{bmatrix} a_{11} & a_{12}\\a_{21} & a_{22} \end{bmatrix} \begin{bmatrix} b_{11} & b_{12}\\b_{21} & b_{22} \end{bmatrix} = \begin{bmatrix} a_{11}b_{11} + a_{12}b_{21} & a_{11}b_{12} + a_{12}b_{22}\\a_{21}b_{11} + a_{22}b_{21} & a_{21}b_{12} + a_{22}b_{22} \end{bmatrix}$$

Reason for this kind of product is that it is useful for the use of matrices as a representation of **linear transforms**. When several linear transforms are performed in succession this corresponds to as many matrix products using the definition above. Linear transformations are for example rotations in plane.

The product is written above  for matrices with two rows and two columns, but the product is calculated in similar way for larger matrices.

Matrix product $AB$ is defined if the left factor $A$ has the same number of columns as the right factor $B$ has rows. 

Generally, the product of $m \times n$ matrix and $n \times p$ matrix will be $m \times p$ matrix. 

For example a $100 \times 20$ matrix and a $20 \times 5$ matrix will have the product of size $100 \times 5$.

a $100 \times 10$ matrix $A$ and a $20 \times 100$ matrix $B$ do not have a product $AB$, but the product $BA$ has the size $20 \times 10$.


Matrix multiplication operator is @. Element-by-element product is *.

````python
a = np.array([[2,43],[4,-3]])
b = np.array([[-2,32],[22,33]])
print("Matrix product a*b =\n",a@b)
print("Matrix product b*a =\n",b@a)
print("Matric product is also np.dot(a,b)= \n",np.dot(a,b))
print("Elementwise product a.*b = \n",a*b)
````

Notice the different results in ```a*b``` and ```b*a```. The matrix product is **not commutative**. So in general for two matrices $$AB \neq BA$$

Only in some special cases $AB = BA$ for matrices. So the order is important in matrix multiplication.

Applications utilizing matrix multiplication are for example 
* [rotations](https://en.wikipedia.org/wiki/Rotation_matrix) and translations in plane or space 
* [resource allocation in economics](https://en.wikipedia.org/wiki/Matrix_multiplication#Matrix_product_.28two_matrices.29)
* linear regression using the normal equations
* optimization algorithms

### Broadcasting

Python for-loops are slow. This is  because of type checking happening in each oteration. To gain efficiency **broadcasting** can be used to eliminate for loops in some case.

Matrix sums and differences are mathematically defined for both operands to have exactly the same dimensions. So the example above violated that rule, but in this case the violation was only practical. Otherwise, the addition would require a for loop for vectors and n loops for n- dimensional arrays. This extension of laws of matrix algebra is one case of bradcasting. Broadcasting means that one operand is cahnged to a different array in "sensible" way. 

Using the broadcasting a vector with `n` elements can be added to an array whose one dimension is `n`. Then values in the vector are extended to a matrix repeating the values so that the dimension mattch.

### Example
Sum of row vector and a matrix 

````python
x = np.array([1,2,3])
A = np.array([[5,6,7],[8,9,10]])
B = A + x #x is extended to [[1,2,3],[1,2,3]]
print("x =",x)
print("A =\n",A)
print("B =\n",B)
````

Sum of a matrix and a column vector.

````python
x = np.array([[2],[4]])
A = np.array([[5,6],[7,8]])
B = A + x #x is extended to [[2,2],[4,4]]
C = x + A
print("x =",x)
print("A =\n",A)
print("B =\n",B )
print("C = \n",C)
````

## Slicing

Slicing means that the `:` operator is used select a portion of the array. Slicing in numpy arrays return a **view** of the array. This is different from lists, where a completely new list is created. 

````python

A = np.array([[[10, 11, 12], [13, 14, 15], [16, 17, 18]],
               [[20, 21, 22], [23, 24, 25], [26, 27, 28]],
               [[30, 31, 32], [33, 34, 35], [36, 37, 38]]])

print(A[:2,1:,:2]) #The first two planes with :2, from there from the second row to the bottom with 1: and the first two rows with :2
B = A[:2,1:,:2]
B[0,0] = 1000 #Will change A also in two locations
print("A=\n",A)
````
