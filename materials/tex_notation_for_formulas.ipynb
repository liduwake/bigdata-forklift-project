# Writing formulas in Tex notation

Tex notation makes it possible to write formulas using ordinary keyboard ans ASCII characters.
A pdf of Tex notation is [here](https://www.cmor-faculty.rice.edu/~heinken/latex/symbols.pdf)

#### Inline and displayed formulas
In Jupyter notebooks the formulas are enclosed using `$ $` or `$$ $$`. Single pair `$ $  indicates **inline style** formulas and double characters **display style** formulas.
#### Example
Inline style formula is on a text line, like this `$a = b + c$` and renderd as $a = b + c$

Display style formula is centered and on its own line `$$x = 2y$$` and shows $$x = 2y$$


## Variable names and basic math

Letters are used to represent variables whose value are not known or constants whose representation as decimal numbers would be impratical to use in formulas. When Tex notation is used these will be in italic font. 

#### 

#### Exponentiation
The power in Tex notation is indicated by `^`. So "two to the power of three" is formed using `$2^3$` which is shown as $2^3$.
#### Variable naming and index notation
Multiplication character between two letters or a number and a letter is **not shown**. For example write $xy$ instead of $x*y$.

In mathematical notation all variables should be represented by **single letters**.  So $ab$ is **not a proper variable name**, because it looks like the product of $a$ and $b$. 
Indices can be used, if necessary. Index notation for three variables could be, for example `$x_1,x_2,x_3$` shown as $x_1,x_2,x_3$


Its is common practise to write function names is normal font. This is achieved using \ before the function name. For example, `$y = \sin(3x)$` is shown  $y = \sin(3x)$ 

Fractions with horizontal bar need the keyword: `\frac{numerator}{denominator}`
For example the expression `(2x+1)/(3x^2 + x + 1)` can be written as `$\frac{2x+1}{3x^2 + x + 1}$` and is displayed as $\frac{2x+1}{3x^2 + x + 1}$. 

## Example 

The sombrero function is defined by the formula `z = sin (sqrt (x^2 + y^2)) / (sqrt (x^2 + y^2))`. Let us write the function in Tex notation to make it look better.

First thing is to look how the parentheses are grouped, because those show the grouping of the parts.
The first part is `sin (sqrt (x^2 + y^2))` and the second one is `(sqrt (x^2 + y^2))`. These are also the numerator and denominator in the expression. 

The numerator is `$\sin(\sqrt{x^2 + y^2})$` and the denominator is `$\sqrt{x^2 + y^2}$`. To form the full expression we need to put these in the empty slots of `\frac{}{}`.
This gives `$z = \frac{\sin(\sqrt{x^2 + y^2})}{\sqrt{x^2 + y^2}}$` and is rendered as $$z = \frac{\sin(\sqrt{x^2 + y^2})}{\sqrt{x^2 + y^2}}$$

## Systems of equations

Simultaneous systems of equations can be written using `align` environment. The equations are separated with a pair of backslashes, `\\`. Usually the equations are aligned at the `=` character and this is shown using the combination `&=`.

The string `$$\begin{align} 2x_1 + 3x_2 &= 3 \\ 5x_1 + 7x_2 = 10 \end{align}$$` forms the pair of equations rendered as $$\begin{align} 2x_1 + 3x_2 &= 3 \\ 5x_1 + 7x_2 &= 10 \end{align}$$

## Matrices

To write a matrix a bit more complicated notation is used: `$A = \begin{bmatrix} 1 & 2 & 3 \\ x & x^2 & x+y \end{bmatrix}$` gives
$A = \begin{bmatrix} 1 & 2 & 3 \\ x & x^2 & x+y \end{bmatrix}$

So the column separator is `&` and the row separator is `\\`.

Row and comlumns vectors are special matrices. So a column vector can be written as `b = \begin{bmatrix} 1\\2\\3 \end{bmatrix}` and a row vector as `c = \begin{bmatrix} 5 & 6 & 7 \end{bmatrix}`.
$$b = \begin{bmatrix} 1\\2\\3 \end{bmatrix}$$
$$c = \begin{bmatrix} 5 & 6 & 7 \end{bmatrix}$$
